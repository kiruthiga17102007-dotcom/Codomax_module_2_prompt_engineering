```python

# Day 6 - Zero-shot and Few-shot Prompting

zero_shot_prompt = """
Classify the following review as Positive or Negative.

Review: The product is excellent!
"""

few_shot_prompt = """
Review: I love this product.
Sentiment: Positive

Review: The product is terrible.
Sentiment: Negative

Review: The product is excellent!
Sentiment:
"""

print("ZERO-SHOT PROMPT")
print(zero_shot_prompt)

print("\nFEW-SHOT PROMPT")
print(few_shot_prompt)
```

```python

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

def get_response(prompt):
    inputs = tokenizer(prompt, return_tensors="pt")
    outputs = model.generate(**inputs, max_new_tokens=50)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

zero_shot = """
Classify this review as Positive or Negative:
The product is excellent!
"""

few_shot = """
Review: I love this product.
Sentiment: Positive

Review: The product is terrible.
Sentiment: Negative

Review: The product is excellent!
Sentiment:
"""

print("ZERO-SHOT RESULT:")
print(get_response(zero_shot))

print("\nFEW-SHOT RESULT:")
print(get_response(few_shot))
```


## Observation

Zero-shot prompting performs a task without examples, whereas few-shot prompting provides examples to guide the model.

Both techniques help us understand how prompt structure influences AI-generated responses.


```python

# Day 7 - Prompt Components

prompts = {
    "Role-based Prompt":
        "You are a Python teacher. Explain Python loops to a beginner.",

    "Context-based Prompt":
        "I am a second-year CSE student. Explain Generative AI in simple English.",

    "Instruction-based Prompt":
        "Explain machine learning in exactly three bullet points.",

    "Constraint-based Prompt":
        "Explain Artificial Intelligence in less than 30 words."
}

for title, prompt in prompts.items():
    print("\n" + title)
    print("Prompt:", prompt)
    print("Response:", get_response(prompt))
    print("-" * 50)
```


## Observation

1. Roles help define the AI's expected behavior.
2. Context provides background information.
3. Instructions clearly describe the required task.
4. Constraints control the format or length of responses.

## Conclusion

Combining roles, context, instructions and constraints helps create more structured and relevant prompts.
